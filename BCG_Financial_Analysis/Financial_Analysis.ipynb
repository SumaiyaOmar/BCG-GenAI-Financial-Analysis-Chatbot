## Data Sources and Methodology

The dataset covers fiscal years 2023, 2024, and 2025 and includes the following financial metrics:

- Total Revenue
- Net Income
- Total Assets
- Total Liabilities

All financial values are recorded in USD millions.

### Methodology

1. **Data Collection:** Extracted financial figures from annual reports and organized them into a structured dataset using Microsoft Excel.

3. **Financial Calculations:** Calculated year-over-year percentage changes in revenue, net income, and operating cash flow, along with net profit margins and liabilities-to-assets ratios.
4. **Trend Analysis:** Compared financial performance across companies and fiscal years using pandas.
5. **Data Visualization:** Created charts using Matplotlib to illustrate revenue trends, profitability, and operating cash flow.
6. **Aggregation:** Used pandas `groupby()` and `agg()` functions to calculate average financial performance for each company.
7. **Interpretation:** Summarized significant trends, differences, and financial findings.

```python
import pandas as pd
```

```python
df = pd.read_csv("Financial_Data_2023_2025.csv")

df
```

## Financial Metrics and Year-over-Year Analysis

To evaluate financial performance, additional metrics were calculated using pandas.

- **Revenue Growth (%):** Measures the percentage change in revenue compared with the previous fiscal year.
- **Net Income Growth (%):** Measures the percentage change in net income compared with the previous fiscal year.
- **Net Profit Margin (%):** Measures the percentage of revenue retained as net income.
- **Liabilities to Assets (%):** Measures the proportion of total assets represented by liabilities.
- **Operating Cash Flow Growth (%):** Measures the percentage change in cash generated from operating activities.

The `groupby()` and `pct_change()` functions were used to calculate year-over-year growth separately for each company.

**Note:** Growth values for 2023 are unavailable (`NaN`) because the dataset does not include financial data for 2022.

```python
df = df.sort_values(["Company", "Year"])

df["Revenue Growth (%)"] = (
    df.groupby("Company")["Total Revenue"]
      .pct_change() * 100
)

df
```

```python
df["Net Income Growth (%)"] = (
    df.groupby("Company")["Net Income"]
      .pct_change() * 100
)

df
```

```python
df["Net Profit Margin (%)"] = (
    df["Net Income"] / df["Total Revenue"]
) * 100

df
```

```python
df["Liabilities to Assets (%)"] = (
    df["Total Liabilities"] / df["Total Assets"]
) * 100

df
```

```python
df["Operating Cash Flow Growth (%)"] = (
    df.groupby("Company")["Operating Cash Flow"]
      .pct_change() * 100
)

df
```

```python
percentage_columns = [
    "Revenue Growth (%)",
    "Net Income Growth (%)",
    "Net Profit Margin (%)",
    "Liabilities to Assets (%)",
    "Operating Cash Flow Growth (%)"
]

df[percentage_columns] = df[percentage_columns].round(2)

df
```

```python
import matplotlib.pyplot as plt

print("Matplotlib is ready!")
```

## Financial Data Visualizations

### Revenue Trends (2023–2025)

This line chart compares the annual revenue of Apple, Microsoft, and Tesla.

Apple consistently generated the highest revenue during the period. However, Microsoft demonstrated the strongest percentage revenue growth from 2023 to 2025, while Tesla's revenue declined slightly overall.

```python
plt.figure(figsize=(10, 6))

for company in df["Company"].unique():
    company_data = df[df["Company"] == company]

    plt.plot(
        company_data["Year"],
        company_data["Total Revenue"],
        marker="o",
        label=company
    )

plt.title("Revenue Trends: Apple vs Microsoft vs Tesla (2023–2025)")
plt.xlabel("Year")
plt.ylabel("Total Revenue (USD Millions)")
plt.xticks([2023, 2024, 2025])
plt.legend()
plt.grid(True)

plt.show()
```

### Net Profit Margin Comparison (2023–2025)

This bar chart compares the net profit margins of Apple, Microsoft, and Tesla over three fiscal years.

Microsoft maintained the highest net profit margin throughout the analysis period, demonstrating consistent profitability relative to revenue.

Apple maintained relatively stable profit margins, while Tesla experienced a significant decline, from 15.47% in 2023 to 4.07% in 2025.

These findings highlight differences in how efficiently the three companies convert revenue into net income.

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

for company in df["Company"].unique():
    company_data = df[df["Company"] == company]

    plt.bar(
        company_data["Year"].astype(str) + "\n" + company,
        company_data["Net Profit Margin (%)"],
        label=company
    )

plt.title("Net Profit Margin Comparison (2023–2025)")
plt.xlabel("Year and Company")
plt.ylabel("Net Profit Margin (%)")
plt.xticks(rotation=45)
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```

### Operating Cash Flow Trends (2023–2025)

This line chart illustrates changes in operating cash flow for Apple, Microsoft, and Tesla between 2023 and 2025.

Microsoft demonstrated the strongest growth in operating cash flow, increasing from $87,582 million in 2023 to $136,162 million in 2025.

Apple maintained relatively stable operating cash flow, although it declined slightly in 2025 compared with 2024.

Tesla generated significantly lower operating cash flow than Apple and Microsoft, with a small overall increase between 2023 and 2025.

These findings demonstrate Microsoft's strong improvement in cash generation from operating activities during the analysis period.

```python
plt.figure(figsize=(10, 6))

for company in df["Company"].unique():
    company_data = df[df["Company"] == company]

    plt.plot(
        company_data["Year"],
        company_data["Operating Cash Flow"],
        marker="o",
        linewidth=2,
        label=company
    )

plt.title("Operating Cash Flow Trends (2023–2025)")
plt.xlabel("Year")
plt.ylabel("Operating Cash Flow (USD Millions)")
plt.xticks([2023, 2024, 2025])
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
company_summary = df.groupby("Company").agg({
    "Total Revenue": "mean",
    "Net Income": "mean",
    "Net Profit Margin (%)": "mean",
    "Operating Cash Flow": "mean"
})

company_summary = company_summary.round(2)

display(company_summary)
```
